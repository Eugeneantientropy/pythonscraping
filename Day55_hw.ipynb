📰 應用案例：自動追蹤新聞網站的新文章
📌 需求場景：
你想每天早上自動抓取某個新聞網站的「最新頭條」，例如科技新聞，並把它們整理成一份報告或推播通知。

✅ 為什麼需要「自動化更新」？
因為新聞每天都在更新，手動去抓太麻煩。用爬蟲 + 自動排程，就可以每天固定時間自動幫你抓資料！

🧠 簡單說明流程：
🕒 設定排程（例如每天早上 8 點）

🌐 爬蟲啟動 → 前往指定的新聞網站

📄 擷取標題、時間、連結等資訊

📬 儲存成報表 / 寄送 Email / Line 通知
