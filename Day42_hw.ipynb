```python
import numpy as np
np.random.seed(888)
import pandas as pd
import plotly.express as px
```

### 折線圖
- 請根據以下資料點繪製折線圖

```python
# 繪圖資料點
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.cos(2 * np.pi * t)

fig = px.line(x=t, y=s, height=600, width=800)
# 設置標題、軸標題
fig.update_layout(
    title="Example Title",
    xaxis_title="X Axis",
    yaxis_title="Y Axis",
    font=dict(
        family="Courier New, monospace",
        size=18,
        color="RebeccaPurple"
    )
)
# 展示圖表
fig.show()
```

### 垂直長條圖
- 請根據以下資料點繪製水平長條圖

```python
# 生成資料點
import seaborn as sns
planets = sns.load_dataset("planets")

#繪製圖形
fig = px.bar(planets, x="method", y="distance", width=800)
fig.show()
```

### 直方圖
- 請根據以下資料點，繪製直方圖


```python
# 生成資料點
planets = sns.load_dataset("planets")

# 繪圖資料點
penguins = sns.load_dataset("penguins")
# 繪製圖形
fig = px.histogram(planets, x="method", nbins=50, width=800)  # nbins 可以調整想要分隔的區間個數
fig.show()
```

### 圓餅圖
- 請根據以下資料點，繪製圓餅圖


```python
import plotly.express as px

# 生成資料點
planets = sns.load_dataset("planets")

fig = px.pie(planets, values='distance', names='method', title='Total bill of different day', height=500, width=500)
fig.show()
```

### 矩陣圖（熱點圖）
- 根據以下隨機生成的相關係數值，繪製矩陣圖

```python
import pandas as pd

# 生成隨機資料
df = pd.DataFrame(np.random.rand(15, 15))
# 計算欄位間相關係數
corr = df.corr().round(2)

fig = px.imshow(corr, text_auto=True, width=600, height=600)
fig.show()
```
