```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(888)

import seaborn as sns
# 設置繪圖風格
sns.set_theme(style="darkgrid")  # {darkgrid, whitegrid, dark, white, ticks}

%matplotlib inline
```

### 折線圖
- 請根據以下資料點繪製折線圖

```python
# 繪圖資料點
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.cos(2 * np.pi * t)
# 設定畫版大小
sns.set(rc={'figure.figsize': (11.7, 8.27)})
# 設定畫板風格
sns.set_theme(style="darkgrid")
# 繪製資料點並展示圖表
sns.lineplot(x=t, y=s)
# 展示圖表
plt.show()

```

### 直方圖
- 請根據以下資料點，繪製直方圖


```python
# 生成資料點
mean = 10
std = 2
x = mean + std * np.random.randn(500)

sns.set(rc={'figure.figsize': (8.27, 9.7)})
sns.set_theme(style="darkgrid")
sns.histplot( data=x)
plt.show()
```

### 圓餅圖
- 請根據以下資料點，繪製圓餅圖


```python
# 生成資料點，包含類別名、資料點、區塊各自和圓心的距離
labels = 'Bulbasaur', 'Charmander', 'Charizard', 'Metapod', 'Beedrill', 'Spearow'
sizes = [318, 309, 312, 329, 298, 305]
explode = (0, 0.1, 0, 0, 0, 0)

colors = sns.color_palette('bright')[:6]
plt.pie(sizes, explode=explode, labels=labels, colors=colors, 
        autopct='%.1f%%', shadow=True, startangle=90)
plt.show()
```

### 矩陣圖（熱點圖）
- 根據以下隨機生成的相關係數值，繪製矩陣圖

```python
import pandas as pd

# 生成隨機資料
df = pd.DataFrame(np.random.rand(15, 15))
# 計算欄位間相關係數
corr = df.corr().round(2)

# 創建畫板及子圖
fig, ax = plt.subplots(figsize=(15, 15))
# 繪製矩陣圖
sns.heatmap(corr)
# 使用 ax.text 遍歷各個位置，標上相關係數數值
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        text = ax.text(i+0.5, j+0.5, corr.values[i, j],
               ha='center', va='center', color='w')
plt.show()
```
