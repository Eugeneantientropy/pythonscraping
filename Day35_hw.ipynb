```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

# 建立瀏覽器（這邊用 Chrome）
driver = webdriver.Chrome()  # 如果 chromedriver 不在環境變數，要指定路徑
driver.maximize_window()  # 視窗放大一點，避免按鈕點不到
```

```python
driver.get("https://www.linkedin.com/login")

# 輸入帳號
username = driver.find_element(By.ID, "username")
username.send_keys("")

# 輸入密碼
password = driver.find_element(By.ID, "password")
password.send_keys("")

# 按下登入
password.send_keys(Keys.RETURN)

time.sleep(3)  # 等待頁面加載
```

```python
# 進入「人脈」頁面
driver.get("https://www.linkedin.com/mynetwork/invite-connect/connections/")

time.sleep(5)  # 等頁面載入聯絡人清單
```

```python
# 自動往下滑動，觸發 LinkedIn 載入更多聯絡人
last_height = driver.execute_script("return document.body.scrollHeight")
while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height
```

```python
# 抓所有聯絡人的連結
contacts = driver.find_elements(By.CSS_SELECTOR, "a.mn-connection-card__link")

profile_links = [link.get_attribute("href") for link in contacts]

for link in profile_links:
    driver.get(link)
    time.sleep(3)  # 等個人頁面載入

    try:
        name = driver.find_element(By.CSS_SELECTOR, "h1.text-heading-xlarge").text
        title = driver.find_element(By.CSS_SELECTOR, "div.text-body-medium.break-words").text
        print("姓名:", name)
        print("職稱:", title)
        print("------")
    except:
        print("無法取得資料")
```
