## 作業目標

* 目前程式中的 proxy_ips 是手動輸入的，請根據 https://free-proxy-list.net/ 寫一個可自動化抓取可用 Proxy 的 proxy_ips。




```python
import pandas as pd
import requests
import random

proxy_ips = []

# 用 pandas 直接解析 HTML 表格
url = 'https://free-proxy-list.net/'
dfs = pd.read_html(requests.get(url).text)

# 取第一個表格（就是 proxy 列表）
df = dfs[0]

# 選出支援 HTTPS 且匿名的代理 IP
for _, row in df.iterrows():
    if row['Https'] == 'yes' and row['Anonymity'] in ['anonymous', 'elite proxy']:
        proxy = f"http://{row['IP Address']}:{row['Port']}"
        proxy_ips.append(proxy)

print(f'共取得 {len(proxy_ips)} 組代理 IP')

# 測試其中 10 組 proxy 是否可用
for i in range(10):
    ip = random.choice(proxy_ips)
    print('Use', ip)
    try:
        resp = requests.get('https://api.ipify.org?format=json',
                            proxies={'http': ip, 'https': ip}, timeout=10).json()
        print(resp)
    except:
        print('Fail')
```
