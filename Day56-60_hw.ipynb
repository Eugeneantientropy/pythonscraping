```python
import requests
from bs4 import BeautifulSoup

url = 'https://www.moneydj.com/kmdj/news/newsreallist.aspx?a=mb010000&page=1'
headers = {
    "User-Agent": "Mozilla/5.0"
}

res = requests.get(url, headers=headers)
soup = BeautifulSoup(res.text, 'html.parser')

# 選出所有 <a>，href 裡有 /newsviewer 就是新聞
anchors = soup.find_all('a', href=True)
for a in anchors:
    if '/kmdj/news/newsviewer.aspx' in a['href']:
        title = a.get('title') or a.text.strip()
        link = "https://www.moneydj.com" + a['href']
        print(f"標題：{title}")
        print(f"連結：{link}")
        print('-' * 40)
```

```python
import requests
from bs4 import BeautifulSoup
import time

# 設定 PTT 的 cookie，表示已滿 18 歲
cookies = {'over18': '1'}

base_url = 'https://www.ptt.cc'

def get_articles_from_index(url):
    res = requests.get(url, cookies=cookies)
    soup = BeautifulSoup(res.text, 'html.parser')
    articles = []

    for item in soup.select('div.r-ent'):
        title_tag = item.select_one('div.title a')
        if not title_tag:
            continue  # 被刪除的文章
        title = title_tag.text.strip()
        link = base_url + title_tag['href']
        articles.append({'title': title, 'link': link})

    return articles

def get_article_detail(url):
    res = requests.get(url, cookies=cookies)
    soup = BeautifulSoup(res.text, 'html.parser')

    # 抓時間
    time_tag = soup.select_one('span.article-meta-tag:-soup-contains("時間")')
    time_text = time_tag.find_next_sibling('span').text if time_tag else '未知'

    # 抓留言
    pushes = []
    for push in soup.select('div.push'):
        tag = push.select_one('span.push-tag').text.strip()
        user = push.select_one('span.push-userid').text.strip()
        content = push.select_one('span.push-content').text.strip()
        pushes.append(f"{tag} {user}: {content}")

    return time_text, pushes

# 抓首頁的文章列表
index_url = 'https://www.ptt.cc/bbs/Stock/index.html'
articles = get_articles_from_index(index_url)

# 顯示每篇文章的時間與留言
for article in articles:
    print(f"文章標題：{article['title']}")
    print(f"文章連結：{article['link']}")
    print(f"發文時間：{time_text}")
    print(f"留言數量：{len(pushes)}")
    print("留言內容：")
    for p in pushes[:5]:  # 只印前 5 則留言，避免太長
        print("    ", p)
    print('-' * 40)
    time.sleep(1)  # 避免對伺服器造成壓力
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# 設定無頭模式
options = Options()
options.add_argument('--headless')
options.add_argument('--disable-gpu')
driver = webdriver.Chrome(options=options)

# Step 1: 開啟分類頁面
driver.get('https://www.pressplay.cc/project?type=507CF74F93AD23B6584EF2C5D5D4430E')
time.sleep(5)  # 等待 JavaScript 載入課程卡片

# Step 2: 模擬滾動或載入更多（這邊可擴充成點按按鈕或滾動到底部）

# Step 3: 抓取所有課程連結
soup = BeautifulSoup(driver.page_source, 'html.parser')
cards = soup.select('a[href^="/project/"]')  # 所有課程連結
course_links = ["https://www.pressplay.cc" + a['href'] for a in cards if '/project/' in a['href']]

# 去除重複課程
course_links = list(set(course_links))

# Step 4: 爬每個課程頁的細節
for url in course_links[:5]:  # 這邊限制前5個測試用，可改成全部
    driver.get(url)
    time.sleep(3)
    soup = BeautifulSoup(driver.page_source, 'html.parser')

    # 課程標題
    title = soup.find('h1').text.strip() if soup.find('h1') else "N/A"

    # 課程描述 / 內容簡介
    intro = soup.find('div', class_='project-intro')
    description = intro.text.strip() if intro else "N/A"

    # 追蹤人數
    follower = soup.find('div', class_='count')
    follower_count = follower.text.strip() if follower else "N/A"

    # 授課者名稱
    author = soup.select_one('div.teacher-name')
    teacher = author.text.strip() if author else "N/A"

    # 收藏數：按 ❤️ 的人數（可擴充）
    # 關鍵字：從 meta tags
    meta_keywords = soup.find('meta', attrs={'name': 'keywords'})
    keywords = meta_keywords['content'] if meta_keywords else "N/A"

    print(f"課程名稱：{title}")
    print(f"簡介：{description[:50]}...")
    print(f"追蹤人數：{follower_count}")
    print(f"老師：{teacher}")
    print(f"關鍵字：{keywords}")
    print(f"課程連結：{url}")
    print("-" * 40)

driver.quit()
```
