```python
import matplotlib.pyplot as plt
# 設置繪圖風格
print(plt.style.available)
# 運行 cell 印出所有可用的風格，從中選擇一個填入下方空格
plt.style.use("classic")

import numpy as np
np.random.seed(888)

%matplotlib inline
```

### `title` 設置圖表標題
- `ax.set_title()`

```python
# 生成隨機資料點
# Hint: 使用 np.random.randn
x = np.random.randn(50)
y = np.random.randn(50)

# 設定畫布
fig, ax = plt.subplots(figsize=(6, 4))

# 繪製散點圖
ax.scatter(x, y)

# 設置標題
ax.set_title("Random Scatter Plot")

# 自動調整版面
fig.tight_layout()

# 顯示圖表（補充）
plt.show()
```

### `legend` 設置圖表圖例
- `ax.legend()`

```python
# 為了讓畫布上有不同顏色的資料點，再生成一份相同數量的 y 軸資料點
# 生成隨機資料點
x = np.random.randn(50)
y1 = x + np.random.randn(50) * 0.5  # 第一組 y 資料（加點雜訊）
y2 = np.random.randn(50)            # 第二組 y 資料（獨立生成）

# 設定畫布
fig, ax = plt.subplots(figsize=(6, 4))

# 繪製兩組散點圖（加上 label）
ax.scatter(x, y1, color='blue', label='Group 1')
ax.scatter(x, y2, color='orange', label='Group 2')

# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")

# 設置圖例
ax.legend(loc="upper right")

# 自動調整排版
fig.tight_layout()

# 顯示圖形
plt.show()
```

### 設置座標軸範圍
- `ax.axis()`
- `ax.set_xlim()` / `ax.set_ylim()`

```python
# 設定畫布
fig, ax = plt.subplots(figsize=(6, 4))
# 繪製散點圖
ax.scatter(x, y)
# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")
# 設置圖例
ax.legend(loc="upper right")
# 將座標移動為 x: -2~4, y: -1~5 的範圍
ax.axis([-2, 4, -1, 5])  # [x_min, x_max, y_min, y_max]
fig.tight_layout()
```

```python
# 設定畫布
fig, ax = plt.subplots(figsize=(6, 4))
# 繪製散點圖
ax.scatter(x, y)
# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")# 設置圖例

ax.legend(loc="upper right")# 將座標移動為 x: -2~4, y: -1~5 的範圍
ax.set_xlim(-2, 4)
ax.set_ylim(-1, 5)
fig.tight_layout()
```

### `label` 設置座標軸標題
- `ax.set_xlabel()` / `ax.set_ylabel()`

```python
# 生成隨機資料
x = np.random.randn(50)
y = np.random.randn(50)

# 設定畫布
fig, ax = plt.subplots(figsize=(6, 4))

# 繪製散點圖（加 label 才能顯示圖例）
ax.scatter(x, y, label="Group 1")

# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")

# 設置圖例
ax.legend(loc="upper right")

# ✅ 設定座標軸顯示範圍
ax.set_xlim(-2, 4)
ax.set_ylim(-1, 5)

# ✅ 設置座標軸標題
ax.set_xlabel("True")
ax.set_ylabel("False")

# 自動排版
fig.tight_layout()

# 顯示圖形
```

### `ticks` 設置座標軸刻度字型、大小
- `ax.set_xticklabels()` / `ax.set_yticklabels()`

```python
# 設定畫布
fig, ax = plt.subplots(figsize=(6, 4))
# 繪製散點圖
ax.scatter(x, y, label="Group 1")
# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")

# 設置圖例
ax.legend(loc="upper right")

# 設置座標軸標題
ax.set_xlabel("This is X axis")
ax.set_ylabel("This is Y axis")
# 設置座標刻度樣式
ax.set_xticklabels([0, -3, -2, -1, 0, 1, 2], color="green", fontsize=12, rotation=45)
ax.set_yticklabels([0, -2, -1, 0, 1, 2], color="blue", fontsize=12, rotation=90)
fig.tight_layout()
```

### `annotate` 設置資料點註釋
- `ax.annotate()`

```python
fig, ax = plt.subplots(figsize=(6, 4))
# 繪製散點圖
ax.scatter(x, y, label="Group 1")
# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")

# 設置圖例
ax.legend(loc="upper right")

# 設置座標軸標題
ax.set_xlabel("This is X axis")
ax.set_ylabel("This is Y axis")
# 設置資料點註釋
# Hint: xy 是註釋要出現的座標，xytext 是文字座標，arrowprops 是箭頭的樣式設定參數，facecolor 是箭頭的顏色
ax.annotate('<annotation_text>', xy =(2, 1), xytext =(4, 1.5),  
            arrowprops = dict(facecolor ='orange'))
fig.tight_layout()
```

### `text` / `arrow` 設置圖表說明文字&箭頭
- `ax.text()` / `ax.arrow()` 

```python
fig, ax = plt.subplots(figsize=(6, 4))
# 繪製散點圖
ax.scatter(x, y, label="Group 1")
# 設置標題
ax.set_title("Random Scatter Plot with Two Groups")

# 設置圖例
ax.legend(loc="upper right")

# 設置座標軸標題
ax.set_xlabel("This is X axis")
ax.set_ylabel("This is Y axis")
# 設置資料點註釋
# Hint: xy 是註釋要出現的座標，xytext 是文字座標，arrowprops 是箭頭的樣式設定參數，facecolor 是箭頭的顏色
ax.annotate('Important Point', xy =(2, 1), xytext =(4, 1.5),  
            arrowprops = dict(facecolor ='orange'))
# 設置圖表說明文字
ax.text(x[0], y[0], "Some text here", ha="left", family='serif', style='italic', size=12, rotation=10, wrap=True)
# 設置箭頭
ax.arrow(5, 5, 2, 2,  # 起始x座標, 起始y座標, 往x座標方向的長度, 往y座標方向的長度
          head_width=0.2, width=0.02, ec="green")  # 箭頭寬度, 箭身寬度, 顏色
fig.tight_layout()
```
