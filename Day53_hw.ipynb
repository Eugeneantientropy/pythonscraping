# 加速：多線程爬蟲
- 了解多線程爬蟲加速原理與實作

## 作業目標

* 找一個之前實作過的爬蟲改用多線程改寫，比較前後時間的差異。





```python
import requests
import time

urls = ["https://httpbin.org/delay/1" for _ in range(5)]  # 模擬 5 個延遲的網址

start_time = time.time()

for url in urls:
    response = requests.get(url)
    print(f"完成: {url}，狀態碼: {response.status_code}")

end_time = time.time()
print(f"單線程總耗時: {end_time - start_time:.2f} 秒")
```

```python
import requests
import time
import threading

urls = ["https://httpbin.org/delay/1" for _ in range(5)]

def fetch(url):
    response = requests.get(url)
    print(f"完成: {url}，狀態碼: {response.status_code}")

start_time = time.time()

threads = []
for url in urls:
    thread = threading.Thread(target=fetch, args=(url,))
    thread.start()
    threads.append(thread)

# 等待所有線程完成
for thread in threads:
    thread.join()

end_time = time.time()
print(f"多線程總耗時: {end_time - start_time:.2f} 秒")
```
