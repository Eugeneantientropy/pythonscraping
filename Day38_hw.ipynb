## 作業說明
- 以簡單範例資料繪圖，練習 Matplotlib 的畫板設置
  - 圖表標題 (title)
  - 圖表圖例 (legend)
  - 座標軸範圍 (axis)
  - 座標軸標題 (label)
  - 座標軸刻度字型、大小等 (ticks)
  - 圖表網格線 (grid)
  - 資料點標籤 (annotation)
  - 在圖表特定位置加上文字或箭頭 (text/arrow)
  

```python
import matplotlib.pyplot as plt
# 設置繪圖風格
print(plt.style.available)
# 運行 cell 印出所有可用的風格，從中選擇一個填入下方空格
# plt.style.use("_____")

import numpy as np
np.random.seed(888)

%matplotlib inline
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 生成隨機資料點
x = np.random.randn(50)
y = np.random.randn(50)

# 設定畫布大小（單位：英吋）
plt.figure(figsize=(3, 5))  # 5 吋寬，8 吋高

# 繪製散點圖
plt.scatter(x, y)

# 顯示圖形
plt.show()
```

### `title` 設置圖表標題
- `plt.title()`

```python
# 設定畫布大小
plt.figure(figsize=(3, 5))  # 5 吋寬，8 吋高
plt.scatter(x, y)
# 設置標題
# Hint: 輸入想要設定為標題的字串
plt.title("Random Scatter Plot")
plt.show();
```

### `legend` 設置圖表圖例
- `plt.legend()`

```python
# 為了讓畫布上有不同顏色的資料點，再生成一份相同數量的 y 軸資料點
# 生成隨機資料點
y1 = y.copy()
y2 = np.random.randn(len(y))  # 👈 隨機產生另一組 y 值

plt.figure(figsize=(8, 5))  # 👈 設定圖的大小

# 繪製散點圖
plt.scatter(x, y1, label="OG")  # label 是圖例上會出現的名稱
plt.scatter(x, y2, label="Random")
plt.title("Random Scatter Plot")  # 👈 圖的標題

# 設置圖例
plt.legend(loc="upper right")  # 👈 設定圖例位置
plt.show()
```

### 設置座標軸範圍
- `plt.axis()`
- `plt.xlim()` / `plt.ylim()`


```python
plt.figure(figsize=(8, 5))  # 設定圖形大小

plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")     # 畫第二組散點圖

plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")             # 顯示圖例在右上角

# 將座標移動為 x: 0-5, y: 0-10 的範圍
plt.axis([0, 5, 0, 10])                   # 設定顯示範圍

plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
# 將座標移動為 x: 0-5, y: 0-10 的範圍
# Hint: plt.xlim([x_min, x_max]) 
# Hint: plt.ylim([y_min, y_max])
             # 設定顯示範圍

plt.xlim([0, 5])
plt.ylim([0, 10])
plt.show();
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
# 關閉座標軸
plt.axis("off")
plt.show();
```

### `label` 設置座標軸標題
- `plt.xlabel()` / `plt.ylabel()`


```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
plt.xlabel("This is Y axis")
plt.ylabel("This is X axis")
plt.show();
```

### `ticks` 設置座標軸刻度字型、大小
- `plt.xticks()` / `plt.yticks()`

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
plt.xlabel("This is Y axis")
plt.ylabel("This is X axis")
# 設置座標刻度樣式
plt.xticks(color="blue", fontsize=10, rotation=45)
plt.yticks(color="green", fontsize=10, rotation=0)
plt.show();
```

### `grid` 是否顯示圖表網格線
(某些繪圖風格已經有網格線，可以透過這個設置選擇是否顯示網格線）
- `plt.grid()`

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
plt.xlabel("This is Y axis")
plt.ylabel("This is X axis")
# 設置座標刻度樣式
plt.xticks(color="blue", fontsize=10, rotation=45)
plt.yticks(color="green", fontsize=10, rotation=0)
# 關閉圖表網格線
plt.grid()
plt.show();
```

### `annotate` 設置資料點註釋
- `plt.annotate()`

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
plt.xlabel("This is Y axis")
plt.ylabel("This is X axis")

# 設置座標刻度樣式
plt.xticks(color="blue", fontsize=10, rotation=45)
plt.yticks(color="green", fontsize=10, rotation=0)

# 設置資料點註釋
plt.annotate('This is a point', xy=(x[0], y1[0]), xytext=(x[0]+0.5, y1[0]+1),  
             arrowprops=dict(facecolor='red'))

plt.show();
```

### `text` / `arrow` 設置圖表說明文字&箭頭
- `plt.text()` / `plt.arrow()`

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y1, label="og")     # 畫第一組散點圖
plt.scatter(x, y2, label="rd")  
plt.title("two set data scatter")             # 加上圖標題
plt.legend(loc="upper right")     
plt.xlabel("This is Y axis")
plt.ylabel("This is X axis")

# 設置座標刻度樣式
plt.xticks(color="blue", fontsize=10, rotation=45)
plt.yticks(color="green", fontsize=10, rotation=0)

# 設置圖表說明文字
plt.text(1, 8, "Some text here", ha="left", family='serif', style='italic', size=12, rotation=10, wrap=True)

# 設置箭頭
plt.arrow(1, 7, 1, -2,  # 起點(1,7)，箭頭長度(x方向+1, y方向-2)
          head_width=0.3, width=0.05, ec="red")  # 箭頭樣式設定

plt.show()
```
